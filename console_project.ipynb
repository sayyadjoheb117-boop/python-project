```python
import json  # Used to store and read data in JSON format
import os   # Used to check if the file exists
```

```python
data_file = "students_info.txt"  # File used to store student data
```

```python

def load_students():
    try:
        with open(data_file, "r") as f:
            return json.load(f)
    except (FileNotFoundError, json.JSONDecodeError, PermissionError):
        return []
```

```python
def save_students(students):
    # This function saves student data into the file

    try:
        with open(data_file, "w") as f:  # Open file in write mode
            json.dump(students, f, indent=4)  # Save Python list as JSON

    except PermissionError:  # Handle permission problem
        print("file cannot be save..!")

    except OSError as e:  # Handle file-related error
        print("File error:", e)

    except Exception as e:  # Handle any other error
        print("something went wrong.....!", e)

```

```python
def create_student():
    # This function creates a new student record

    try:
        roll_no = int(input("Enter Roll Number: "))

        if roll_no <= 0:
            print("Please enter a positive value.......!")
            return

        name = input("Enter Student Name: ")

        if "  " in name:
            print("Please enter only one space between words!")
            return

        age = int(input("Enter Student Age: "))

        if age <= 0:
            print("Please enter a positive age!..........")
            return

        course = input("Enter Course Name: ")

        if not name.strip():
            print("Student name cannot be empty!")
            return

        students = load_students()

        for student in students:
            if student["roll_no"] == roll_no:
                print("Roll Number already exists!")
                return

        new_student = {
            "roll_no": roll_no,
            "name": name,
            "age": age,
            "course": course
        }

        students.append(new_student)
        save_students(students)

        print("Student Record Saved Successfully!!")

    except ValueError:
        print("Please enter numbers only!")

    except Exception as e:
        print("Unexpected error:", e)
```

```python
def read_students():
    # This function displays all student records

    try:
        students = load_students()  # Load student data

        if not students:  # Check if list is empty
            print("No student records found!")
            return  # Stop the function

        for ind, student in enumerate(students, start=1):  # Read each student
            print(f"\nStudent {ind}:")  # Print student number
            print(f"Roll Number : {student['roll_no']}")  # Print roll number
            print(f"Name        : {student['name']}")  # Print name
            print(f"Age         : {student['age']}")  # Print age
            print(f"Course      : {student['course']}")  # Print course

    except KeyError as e:  # Handle missing data
        print("Student data is missing:", e)

    except Exception as e:  # Handle other errors
        print("Unexpected error:", e)
```

```python
def update_student():
    # This function updates an existing student

    try:
        students = load_students()  # Load student data

        if not students:  # Check if there are no records
            print("No records to update!")
            return  # Stop the function

        roll_no = int(input("Enter Roll Number to Update: "))  # Take roll number

        if roll_no <= 0:  # Check if roll number is negative or zero
            print("Please enter a positive value!")  # Show error message
            return  # Stop function

        found = False  # Used to check if student is found

        for student in students:  # Check each student
            if student["roll_no"] == roll_no:  # Find student

                new_name = input("Enter New Name: ")  # Take new name
                new_age = int(input("Enter New Age: "))  # Take new age
                new_course = input("Enter New Course: ")  # Take new course

                if not new_name.strip():  # Check if name is empty
                    print("Name cannot be empty!")
                    return  # Stop function

                if new_age <= 0:  # Check if age is negative or zero
                    print("Please enter a positive age!")
                    return  # Stop function

                student["name"] = new_name  # Update name
                student["age"] = new_age  # Update age
                student["course"] = new_course  # Update course

                found = True  # Student was found
                break  # Stop the loop

        if found:  # Check if student was found
            save_students(students)  # Save updated data
            print("Student Details Updated Successfully!!")

        else:
            print("Student Roll Number Not Found!")

    except ValueError:  # Handle non-numeric input
        print("Please enter numbers only!")

    except Exception as e:  # Handle other errors
        print("Unexpected error:", e)
```

```python
def delete_student():
    # This function deletes a student record

    try:
        students = load_students()  # Load student data

        if not students:  # Check if there are no records
            print("No records to delete!")
            return  # Stop the function

        roll_no = int(input("Enter Roll Number to Delete: "))  # Take roll number

        if roll_no <= 0:  # Check if roll number is negative or zero
            print("Please enter a positive value!")  # Show error message
            return  # Stop function

        new_students = [
            student for student in students
            if student["roll_no"] != roll_no
        ]  # Create list without selected student

        if len(new_students) < len(students):  # Check if student was deleted
            save_students(new_students)  # Save new student list
            print("Student Record Deleted Successfully!!")

        else:
            print("Student Roll Number Not Found!")

    except ValueError:  # Handle non-numeric input
        print("Please enter numbers only!")

    except Exception as e:  # Handle other errors
        print("Unexpected error:", e)
```

```python
def menu():
    # This function displays the main menu

    while True:  # Keep menu running until user exits

        try:
            print("\n================================")
            print("   Student Management System")
            print("================================")

            print("1: Create Student Record")
            print("2: View Student Records")
            print("3: Update Student Record")
            print("4: Delete Student Record")
            print("5: Exit")

            choice = input("\nEnter your choice: ")  # Take menu choice

            if choice == "1":  # Check if choice is 1
                create_student()  # Call create function

            elif choice == "2":  # Check if choice is 2
                read_students()  # Call read function

            elif choice == "3":  # Check if choice is 3
                update_student()  # Call update function

            elif choice == "4":  # Check if choice is 4
                delete_student()  # Call delete function

            elif choice == "5":  # Check if choice is 5
                print("Thank You!!")  # Show thank you message
                break  # Stop the program

            else:  # Handle invalid menu choice
                print("Invalid Choice! Please enter 1 to 5.")

        except KeyboardInterrupt:  # Runs when user presses Ctrl + C
            print("\nProgram interrupted by user.")  # Show message to user
            break  # Stop the program

        except Exception as e:  # Runs when any unexpected error occurs
            print("Unexpected error:", e)  # Show the error message
```

```python
menu()
```
